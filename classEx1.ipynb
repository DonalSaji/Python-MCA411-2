```python
import random
answer = random.randint(1, 100)
win=False
while not win:
    guess = int(input("Guess a number between 1 and 100: "))
    if guess < answer:
        print("Too low!")
    elif guess > answer:
        print("Too high!")
    else:
        print("Congratulations! You've guessed the correct number.")
        win=True
```
